# **The Superposition and Coin Analogy: Phase and Interference on Real Quantum Hardware**

###**Superposition:**
The state of the system is not represented by a single statevector, it can be represented by the superposition of two or more statevectors.

**For example:** A qubit is a fundamental quantum mechanical system that has two-dimensional state space. Suppose $|0\rangle$ and $|1\rangle$ form an orthonormal basis for that state space. Then an arbitrary statevector in that state space is given by:
$$\begin{align}
|\psi\rangle = a|0\rangle + b|1\rangle
\end{align}$$

 A state space of a system is any complex vector space with inner product associated with any isolated physical system (Hilbert space).

Intuitively, the states $|0\rangle$ and $|1\rangle$ are analogous to two values 0 and 1 which a bit may take. The way a qubit is different from a bit is that ***superpositions*** of these two states can also exist. So it isn't possible to say that the qubit is definitely in state $|0\rangle$ or $|1\rangle$.


In general, according to the ***"superposition principle: "***

> If individual states are permissible physical states of a system, then the linear superpositions of these states $|\psi_i\rangle$ gives a new state $|\psi\rangle$ that represents a possible physical state of the system.

$$\begin{align}|\psi\rangle = \sum_{i} a_i |\psi_{i}\rangle \end{align}$$

where $a_i$ are the complex numbers and the quantity:

$$\begin{align*} P = \left|\sum_{i} a_i |\psi_i\rangle\right|^2 = \left|\sum_{i} a_i\right|^2 \end{align*}$$
represents the probability for this superposition.

### **Classical Coin**
Tossing a classical coin has a 50% chance of landing heads up or heads down. The classical coin flips are practically random and one can't predict the outcome without knowing exact initial conditions. That's why they are deterministic and can be used as classical example of probabilistic state, given by:

 $$\begin{align}S(coin) = \frac{1}{{2}}|up\rangle + \frac{1}{{2}}|down\rangle\end{align}$$
  where the coefficients represent the probabilities of each of the outcomes.

If I flip a coin 1000 times and record the number of heads up and heads down, then I will get following results:

```python
flips = 1000
flipList = [random.randint(0, 1) for f in range(flips)]

# plotting
plt.hist(flipList)
plt.show()
```

###**Quantum Coin:**
I can create a "quantum coin" by starting a qubit in state $|0\rangle$ and applying a `Hadamard gate`, which places it into equal superposition of $|0\rangle$ and $|1\rangle$. Like the classical coin flip, a qubit can also be measured in two possible states $|0\rangle$ and $|1\rangle$ as I discussed earlier. When I measure this qubit, there will be 50% chance of measuring $|0\rangle$ and 50% chance of measuring $|1\rangle$ which is same as classical coin flip does.

However, there is a difference between mathematics of quantum coin and classical coin. For quantum coin I have:

$$\begin{align*} |\psi\rangle = \frac{1}{\sqrt{2}}|0\rangle + \frac{1}{\sqrt{2}}|1\rangle\end{align*}$$

Here the coefficients of two states are no longer equal to probabilities as in classical states. Instead, it is the square of the coefficients that gives us the probabilities and each of these coefficients can now be complex, meaning they can have both real and imaginary parts.

Despite these differences, the outcome of measuring this state is essentially the same as flipping a coin.

```python
from qiskit import QuantumCircuit
qcoin = QuantumCircuit(1)
qcoin.h(0)
qcoin.measure_all()

qcoin.draw("mpl")
```

So in effect, applying the Hadamard gate to qubit is analogue to flipping a coin.
Just like I flipped a classical coin 1000 times to look at the statistics of the coin landing heads up or down, I will do the same thing with quantum coin using Qiskit. This is done by using a primitive called `Sampler` which will repeat a circuit several times to sample the statistics of the resulting state.

### **The Quantum Coin: Exploring Superposition in Three Dimensions**
Consider I flip a coin in air, and rather than let it fall to the ground, I have the coordination to clap as it passes between hands and sandwich it between my palms. Now, instead of the coin being heads up or down, it's head left or right. The probability will still be 50-50. It shouldn't matter along which dimension I choose to measure the outcome of the classical coin flip.

But a quantum coin behave differently depending on how I measure it. The standard basis measurements on the quantum computer are along the vertical axis (just like for the case of heads up or down). But I can also change the dimension of quantum coin. To measure the quantum coin in x-axis, I consider states $|+\rangle$ and $|-\rangle$.

While `Sampler` primitive only samples in the measurement basis Z, I use `Estimator` primitive to get the expectation value of X. The values for X are:
* +1 for $|+\rangle$
* -1 for $|-\rangle$

If both outcomes have 50-50 probability for the case when quantum coin behave like classical coin, the expectation value that `Estimator` would return is:

$$\begin{align} (0.5)(+1) + (0.5)(-1) =  0\end{align}$$
So for this case, `Estimator` would return an expectation value of 0.

Suppose I have a quantum coin and now, I'm going to measure it along X axis to see the result:

```python
from qiskit.quantum_info import Pauli

# Mapping the problem
qcoin1 = QuantumCircuit(1)
qcoin1.h(0)

obs = Pauli("X")

# Transpiling the Circuit
pm = generate_preset_pass_manager(target = target, optimization_level = 3)
qc_isa = pm.run(qcoin1)

obs_isa =  obs.apply_layout(layout = qc_isa.layout)

# Executing on real quantum hardware
estimator = Estimator(mode = backend)
job = estimator.run([[qc_isa, obs_isa]])

result = job.result()

# Returning the result in classical form and analyzing them.

print(result[0].data.evs)
```

The expectation value of X is 1 which means that qubit is definitely in the $|+\rangle$ state when measured in the X basis:
* Probability of $|+\rangle$ = 100%
* probability of $|-\rangle$ = 0%

So, there is not a 50-50 probability  of measuring $|+\rangle$ and $|-\rangle$. The key idea is that a quantum state can have a random outcomes along one measurement axis but completely predictable outcomes along another axis.

### **Quantum Phase: The Hidden Dimension of Superposition**
The main difference between a classical coin and a quantum coin is "phase". In classical probabilistic state:

$$\begin{align}S(coin) = c_1|up\rangle + c_2 |down\rangle\end{align}$$

Each coefficient is simply a real, positive number that represents the probability of measuring a particular state.

Whereas, a quantum state can be written as:
$$\begin{align} |\psi\rangle = c_1|0\rangle + c_2|1\rangle\end{align}$$
Unlike classical probabilities, the coefficients $c_1$ and $c_2$ are complex numbers. Each coefficient $c_i$ can be expressed as two dimensional vector on complex plane with :
* magnitude $|c_i|$ related to the probability amplitude
* phase / angle $\phi_i$ that it makes with real axis:
$$\begin{align} c_i = |c_i|e^{i\phi_i}\end{align}$$
Phase $\phi_i$ determines how the amplitudes interact.

The phase is important because quantum amplitudes can interfere constructively or destructively like wave.
* If two quantum states are in phase, then their amplitudes add together giving rise to **constructive interference.**
* If two quantum states are out of phase, then their amplitudes cancel each other and as a result, they give rise to **destructive interference.**

But only **relative phase $\Delta\phi = \phi_2 - \phi_1$ between them matters** because it determines how the two quantum states interfere, changing the probability measurements.

The key idea is simple:
> Superposition creates multiple probability amplitudes. While quantum phase controls how probability amplitudes interfere, allowing quantum states to reinforce or cancel each other.

Applying `Hadamard gate` twice demonstrates this idea. Classically, it doesn't make any sense because it is equivalent to flipping a coin that is already flipping. But quantum mechanically, two quantum operations can interact through interference and return the qubit to a definite state.

Let's see what happens to quantum coin:

```python
qcoin2 = QuantumCircuit(1)
qcoin2.h(0)
qcoin2.h(0)
qcoin2.measure_all()

qcoin2.draw("mpl")
```

Starting with state $|0\rangle$, the first `Hadamard gate` transforms the state into superposition as given below:
$$\begin{align}H|0\rangle = \frac{1}{\sqrt{2}}|0\rangle + \frac{1}{\sqrt{2}}|1\rangle\end{align}$$
In this case, both coefficients are fully real and positive such that phase $\phi$ = 0.

When second `Hadamard gate` is applied, then the result is again in superposition of 0 and 1 but with phase $\phi$ = $\pi$ indicated by negative sign:
$$\begin{align} H|1\rangle = \frac{1}{\sqrt{2}}|0\rangle - \frac{1}{\sqrt{2}}|1\rangle\end{align}$$

When these two resulting states combine, the $|1\rangle$ terms cancel by destructive interference, while the $|0\rangle$ terms interfere constructively. The key idea is that second `Hadamard gate` uses phase to create interference, returning the qubit to the definite state $|0\rangle$ from where I started.

Now, I'm going to check my prediction with `Sampler`:

```python
# Transpile
target = backend.target
pass_manager = generate_preset_pass_manager(target = target, optimization_level = 3)
qc_isa = pass_manager.run(qcoin2)

# Executing or Sampling
sampler = Sampler(mode = backend)
job = sampler.run([qc_isa], shots = 1000)
res = job.result()

counts = res[0].data.meas.get_counts()
plot_histogram(counts)
```

Hence it is proved that `Hadamard gate` has canceled out the first state and I have returned to the 0 state from where I started.

Now I will change phase using a "PHASE gate". This is done by first applying `Hadamard gate` to create the superposition, then apply a phase shift of $\pi$ radians, and finally apply the second `Hadamard gate`:

```python
qcoin_p = QuantumCircuit(1)
qcoin_p.h(0)
qcoin_p.p(np.pi, 0)
qcoin_p.h(0)
qcoin_p.measure_all()

qcoin_p.draw('mpl')
```

```python
# Transpile
target = backend.target
pm_p = generate_preset_pass_manager(target = target, optimization_level = 3)
qc_isa_p = pm_p.run(qcoin_p)

# Executing or Sampling
sampler = Sampler(mode = backend)
job = sampler.run([qc_isa_p], shots = 1000)
res = job.result()

# Analyzing
counts = res[0].data.meas.get_counts()
plot_histogram(counts)
```

And now the qubit is measured to be in 1 state instead of 0.
Quantum phase may not be visible in a single measurement, but it can strongly affect later operations through interference.

The same quantum gate can produce completely different results depending on qubit's phase.

### **A Better Analogy for Quantum Superposition: The $\sqrt{NOT}$ Gate**
I'm going to perform another thought experiment with a coin by considering a coin sitting on the table, heads up. To make it heads down, I can simply flip it over which is computational equivalent of a "NOT" gate. This NOT gate is just a $180^\circ$ rotation. Therefore $\sqrt{NOT}$ gate can be visualized as rotating it only $90^\circ$.

Imagine after applying the $\sqrt{NOT}$ gate, the coin stands on its edge. The result is neither heads-up nor heads-down when we measure it, instead the coin will collapse either to heads-up or heads-down with equal probability. But in practice, it should be difficult to predict which side the coin will fall, so it is essentially random.

**In Three Dimensions:**

I can actually measure this coin along three different axes: x, y and z. The $\sqrt{NOT}$ gate puts the quantum coin into a state analogous to coin standing on its edge:
* Along X and Z axes, measurement gives 50:50 random outcomes i.e. equal probability
* Along the y axis, the result is always flat with its head facing the y-direction called "heads forward."

This is similar to what I did in previous section of this project called quantum coin (qubit) in superposition in three dimensions: random in one basis but definite in another.
Therefore, the edge-standing coin is a better analogy for quantum superposition than a coin simply flipping in the air.

I'm going to test this by applying $\sqrt{NOT}$ gate to $|0\rangle$ and using `Estimator` to measure X, Y, and Z.

```python
qcoin_sx = QuantumCircuit(1)
qcoin_sx.sx(0)
qcoin_sx.draw("mpl")
```

```python
obs1 = Pauli("X")
obs2 = Pauli("Y")
obs3 = Pauli("Z")

# Transpiling
pm_sx = generate_preset_pass_manager(target = target, optimization_level = 3)
qc_isa_sx = pm_sx.run(qcoin_sx)
obs1_isa_sx = obs1.apply_layout(layout = qc_isa_sx.layout)
obs2_isa_sx = obs2.apply_layout(layout = qc_isa_sx.layout)
obs3_isa_sx = obs3.apply_layout(layout = qc_isa_sx.layout)

# Executing / running on real hardware
estimator = Estimator(mode = backend)
job = estimator.run([(qc_isa_sx, [obs1_isa_sx, obs2_isa_sx, obs3_isa_sx])])
res = job.result()

# Returning result in classical form and analyzing result
print(res[0].data.evs)
```

The expectation values are 0, -1, 0 for X, Y, and Z respectively. We can also represent these results as:
* X: 50/50
* Y: always -Y
* Z: 50/50

This matches the classical coin rotated by $90^\circ$ using $\sqrt{NOT}$ gate.

We can visualize a qubit as a vector pointing in the direction of coin's head:
* $|0\rangle$  $-->$ points up
* $|1\rangle$ $-->$ points down
* Any equal superposition of $|0\rangle$ and $|1\rangle$ points horizontally

The phase of the state determines the exact horizontal direction which means that the direction it points along the equator is equal to the direction the coefficient $c_2$ points in the complex plane.

### **Visualization of Gates using Bloch Sphere**
Now, every gate that I have used in this project can be visualized  as a rotation of coin or vector:
* NOT: $180^\circ$ around the x-axis
* $\sqrt{NOT}$: $90^\circ$ around the x-axis
* PHASE: rotation of $\phi$ around z-axis
* Hadamard: It is not an uncontrolled coin flip. It is a controlled, deterministic rotation of coin like other gates. It is performed by first rotating $90^\circ$ around the Y-axis, then rotating $180^\circ$ around X-axis.

So there is nothing random about quantum superposition state or any of the standard operation I perform on the qubit. Every operation is deterministic and reversible. The randomness will be only involved when anyone wants to measure the quantum state.

### **Conclusion:**
* I started by comparing the randomness of measuring a quantum superposition state to a coin flip. Through a series of experiments, I learned that there are fundamental differences between a classical coin flip and a quantum superposition.

* I learned that a qubit in superposition is not simply a randomly flipping coin. It can be particularly visualized as "3D Bloch vector" pointing in definite direction, representing quantum state of a qubit.

* Quantum gates rotate this vector or coin deterministically and reversibly. The randomness appears only when I measure the qubit.

* If a state is in superposition, then it depends on the measurement basis. This is because a state can be a superposition in one basis but definite in another basis. Therefore if someone says a qubit in superposition, so he or she must also answer the question, "Superposition of What?".

* But it is important to point out that the coin is still an analogy and Bloch vector is just a visualization tool to calculate the probabilities of measurement outcomes. Bloch vector doesn't tell us about what the quantum state is "really doing" before measurement.

In short, the core idea is:

>Superposition is basis-dependent, quantum evolution is deterministic, and measurement produces the exact randomness.

### **Analogy's Limitation:**
The coin flip analogy is useful for building intuition, but it has a real limit. A classical coin is always heads up or down. Even in mid air, anyone can't be able to tell that in which way the coin will land.

Whereas quantum coin(qubit) in superposition does not behave like classical coin because it doesn't have a hidden, definite value waiting to be revealed. The outcome genuinely doesn't exist until the act of measurement forces it into one state or the other. This randomness only appears at the moment of measurement, not before. So the coin is helpful starting point but it should not be mistaken for how a qubit actually works.
